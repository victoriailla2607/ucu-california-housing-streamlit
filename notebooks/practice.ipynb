# California Housing: análisis previo
Datos censales de California de 1990. Cada observación es un grupo de bloques. El análisis es descriptivo; no estima precios actuales.

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
import plotly.express as px

california = fetch_california_housing()
df_california = pd.DataFrame(california.data, columns=california.feature_names)
df_california["MedHouseVal"] = california.target
df_california.head()
```

```python
print("Dimensiones:", df_california.shape)
print(df_california.dtypes)
print("Faltantes por columna:")
print(df_california.isna().sum())
df_california.describe()
```

## Filtros combinados
Los límites son inclusivos. Los valores siguientes son un ejemplo reproducible, independiente de los widgets.

```python
edad_min, edad_max = 10, 30
latitud_minima = 36.0
df_filtrado = df_california.loc[df_california.HouseAge.between(edad_min, edad_max) & df_california.Latitude.ge(latitud_minima)].copy()
print("Grupos seleccionados:", len(df_filtrado))
if not df_filtrado.empty:
    print("Mediana:", df_filtrado.MedHouseVal.median())
    print("Rango:", df_filtrado.MedHouseVal.max() - df_filtrado.MedHouseVal.min())
else:
    print("Sin resultados")
```

```python
px.histogram(df_filtrado, x="MedHouseVal", nbins=40, title="Distribución filtrada del valor de vivienda", labels={"MedHouseVal": "Valor mediano (100.000 USD)"}).show()
```

```python
px.scatter(df_filtrado, x="MedInc", y="MedHouseVal", opacity=0.35, title="Ingresos y valor de vivienda", labels={"MedInc": "Ingreso mediano (10.000 USD)", "MedHouseVal": "Valor mediano (100.000 USD)"}).show()
```

## Interpretación y límites
MedHouseVal se expresa en cientos de miles de dólares y MedInc en decenas de miles. El target tiene un límite superior cercano a 5,00001. Las asociaciones no demuestran causalidad. La aplicación usa la totalidad de registros seleccionados.
